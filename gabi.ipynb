## Heatmap of Price Tier and Star Count

```python
import altair as alt

# pull cleaned data from project file
%run project.ipynb
```

```python
# limit data to only awarded restaurants, removing selected restuarants (no stars, no BG) 
heat_df = usa[((usa['Star Count'] > 0) | (usa['BibGourmand'] == 1))].copy()

# combine star count and bib gourmand back into one award label
award_map = {0: 'Bib Gourmand', 1: '1 Star', 2: '2 Stars', 3: '3 Stars'}
heat_df['Award'] = heat_df['Star Count'].map(award_map)

# limit the data to only what data needed, award and price tier
heat_df = heat_df[['Award', 'Price Tier']]

heat_df.head()
```

```python
# order awards so bib gourmand is at the bottom and 3 stars at the top
award_order = ['3 Stars', '2 Stars', '1 Star', 'Bib Gourmand']

heatmap = alt.Chart(heat_df).mark_rect(stroke='white').encode(
    x=alt.X('Price Tier:O'),
    y=alt.Y('Award:O', sort=award_order),
    color=alt.Color('count():Q')
).properties(
    width=200,
    height=200,
    title='Michelin Awards by Price Tier (US)'
)

heatmap
```
